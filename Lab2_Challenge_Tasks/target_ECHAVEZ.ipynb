```python
target_name = "ECHAVEZ"
SEED_NUM = 1
redacted_name = ""

for char in target_name:
    if char in "AEIOU":
        redacted_name += str(SEED_NUM)
    else:
        redacted_name += char

print(f"Original Surname:", target_name)
print("SEED_NUM Used:", SEED_NUM)
print("Redacted Output:", redacted_name)
```

```python
engineer_name = "ECHAVEZ"
SEED_NUM = 1
loops = len(engineer_name) # Total cycles from Task 1

# Logic
first_letter_ascii = ord(engineer_name[0])
verification_hash = (first_letter_ascii * SEED_NUM) - loops

print(f"First Letter ASCII (A): {first_letter_ascii}")
print(f"SEED_NUM (B): {SEED_NUM}")
print(f"Total Loops (C): {loops}")
print(f"Calculation: ({first_letter_ascii} * {SEED_NUM}) - {loops}")
print(f"Verification Hash: {verification_hash}")
```
